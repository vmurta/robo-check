```python
import numpy as np
import trimesh

# Vertices (all 6 points)
vertices = np.array([
    [57.052299, 32.159775,  3.447629],
    [24.732647,  0.035505, -4.747519],
    [56.382610, 32.775673,  3.674499],
    [58.734921, 22.388149,  0.952118],
    [12.434930, 23.325809,  0.952111],
    [12.434930, 22.388149,  0.952118],
])

# Each row is a triangle (indices into vertices)
tri_1 = np.array([
    [0, 1, 2]

])

tri_2 = np.array([
    [3, 4, 5]
])
# Create mesh
tri_1_mesh = trimesh.Trimesh(vertices=vertices, faces=tri_1, face_colors=[255, 0, 0, 255])
tri_2_mesh = trimesh.Trimesh(vertices=vertices, faces=tri_2, face_colors=[0, 255, 0, 255])

scene = trimesh.Scene([tri_1_mesh, tri_2_mesh])
scene.show()
```

```python
import open3d as o3d
import numpy as np

vertices = np.array([
    [57.052299, 32.159775,  3.447629],
    [24.732647,  0.035505, -4.747519],
    [56.382610, 32.775673,  3.674499],
    [58.734921, 22.388149,  0.952118],
    [12.434930, 23.325809,  0.952111],
    [12.434930, 22.388149,  0.952118],
])

triangles = np.array([
    [0,1,2],
    [3,4,5]
])

mesh = o3d.geometry.TriangleMesh()
mesh.vertices = o3d.utility.Vector3dVector(vertices)
mesh.triangles = o3d.utility.Vector3iVector(triangles)

mesh.compute_vertex_normals()

o3d.visualization.draw_geometries([mesh])
```
